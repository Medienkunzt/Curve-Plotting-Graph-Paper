```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator
from matplotlib.backends.backend_pdf import PdfPages

# mm -> inch
A4_WIDTH_MM = 296.3
A4_HEIGHT_MM = 210.2
MM_TO_INCH = 1 / 25.4

# XY-Limits
x_max = 4.2
y_max = 3.0

# init plot
fig, ax = plt.subplots(figsize=(A4_WIDTH_MM * MM_TO_INCH, A4_HEIGHT_MM * MM_TO_INCH), dpi=300)
ax2 = ax.twiny()  # second x-axis
ax3 = ax.twiny()  # third x-axis

# set limits
ax.set_xlim(0, x_max)
ax.set_ylim(0, y_max)
ax2.set_xlim(0, x_max)
ax3.set_xlim(0, x_max)

# set title
ax.set_title("Curve Plotting Graph Paper", fontsize=16, fontweight='bold', fontname='Bahnschrift')

# set grid
ax.xaxis.set_major_locator(MultipleLocator(0.3))  # Hauptlinien alle 0.3 (15mm)
ax.xaxis.set_minor_locator(MultipleLocator(0.1))  # Nebenlinien alle 0.1 (5mm)
ax.yaxis.set_major_locator(MultipleLocator(0.2))  # Hauptlinien alle 0.2 (10mm)
ax.yaxis.set_minor_locator(MultipleLocator(0.1))  # Nebenlinien alle 0.1 (5mm)
ax.grid(which='major', linestyle='-', linewidth=0.6, color='black')
ax.xaxis.grid(which='minor', linestyle=(3, (6, 45)), linewidth=0.6, color='black')
ax.yaxis.grid(which='minor', linestyle=(3, (6, 87)), linewidth=0.5, color='black')

# set ticks and labels
ax.set_xticks(np.arange(0, x_max + 0.3, 0.3))
ax.set_xticklabels([f"{x:.1f}" for x in np.arange(0, x_max + 0.3, 0.3)])
ax.set_yticks(np.arange(0, y_max + 0.2, 0.2))
ax.set_yticklabels([f"{y:.1f}" for y in np.arange(0, y_max + 0.2, 0.2)])

# conf major and minor ticks
ax2.xaxis.set_major_locator(MultipleLocator(0.15))
ax2.xaxis.grid(which='major', linestyle='--', linewidth=0.4, color='black')
ax2.xaxis.set_minor_locator(MultipleLocator(0.02))
ax2.xaxis.grid(which='minor', linestyle=(3, (6, 70.44)), linewidth=0.4, color='black')

# conf major and minor ticks
ax2.xaxis.set_ticks_position("bottom")
ax2.xaxis.set_label_position("bottom")
ax2.spines["bottom"].set_position(("axes", -0.08))
ax2.set_frame_on(True)
ax2.patch.set_visible(False)
ax2.set_xticks(np.arange(0, x_max, 0.15))
ax2.set_xticklabels([f"{x:.0f}" if x >= 0 else '' for x in np.arange(21, -8, -1)])
ax2.set_xlabel("Step Table")

# conf major and minor ticks
ax3.xaxis.set_ticks_position("bottom")
ax3.xaxis.set_label_position("bottom")
ax3.spines["bottom"].set_position(("axes", -0.16))
ax3.set_frame_on(True)
ax3.patch.set_visible(False)
ax3.set_xticks(np.arange(0, x_max, 0.3))
ax3.set_xticklabels([f"{x}" if x <= 10 else '' for x in np.arange(0, 15, 1)])
ax3.set_xlabel("Zone Table")

# set axis labels
ax.set_xlabel("LOG EXPOSURE (MILLILUX SECONDS)")
ax.set_ylabel("Density")

# remove spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)

# save plot
pdf_filename = "../GraphPapers/CurvePlottingGraphPaperV0.pdf"
with PdfPages(pdf_filename) as pdf:
    pdf.savefig(fig, bbox_inches='tight')

# show plot
plt.show()
```
